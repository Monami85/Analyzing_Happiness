# Consistency Checks

# 1. Missing Values

```python
# Check for missing values in each column
print(df_time_use.isnull().sum())
```

```python
# Check for missing values in each column
print(df_income_indicator.isnull().sum())
```

```python
# Make a copy of the original dataset
df_income_cleaned = df_income_indicator.copy()

# Fill missing values
df_income_cleaned['IncomeGroup'] = df_income_cleaned['IncomeGroup'].fillna("Not Classified")

# Drop the 'SpecialNotes' and 'Region' columns as they are not needed

df_income_cleaned = df_income_cleaned.drop(columns=['SpecialNotes', 'Region'])

# Verify if missing values are handled
print(df_income_cleaned.isnull().sum())
```

# 2. Inspect Data Types

```python
# Make a copy of df_time_use
df_time_cleaned = df_time_use.copy()

# Check data types of df_time_cleaned
print("Data types in df_time_cleaned :")
print(df_time_cleaned.dtypes)

print("\n------------------------\n")

# Check data types of df_income_cleaned
print("Data types in df_income_cleaned :")
print(df_income_cleaned.dtypes)
```

# 3. Remove Duplicate Rows if any

```python
# Remove duplicate rows from df_time_cleaned
df_time_clean = df_time_cleaned.drop_duplicates()

# Remove duplicate rows from df_income_cleaned
df_income_clean = df_income_cleaned.drop_duplicates()

# Verify if there are any duplicates left
print("Duplicates in df_time_clean :", df_time_clean.duplicated().sum())
print("Duplicates in df_income_clean :", df_income_clean.duplicated().sum())
```

# 4. Rename columns

```python
# Rename the 'Country' column to 'Country name' in df_time_clean
df_time_clean = df_time_clean.rename(columns={'Country': 'Country name'})
```

```python
# Rename the 'TableName' column to 'Country name' in df_income_clean
df_income_clean = df_income_clean.rename(columns={'TableName': 'Country name'})
```

# 5. Check for Outliers

```python
# Summary statistics to identify outliers
print(df_time_clean.describe())
```

```python
# Calculate the IQR for 'Minutes_per_day'
Q1 = df_time_clean['Minutes_per_day'].quantile(0.25)
Q3 = df_time_clean['Minutes_per_day'].quantile(0.75)
IQR = Q3 - Q1

# Define the lower and upper bounds for outliers
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Filter the outliers
outliers = df_time_clean[(df_time_clean['Minutes_per_day'] < lower_bound) | (df_time_clean['Minutes_per_day'] > upper_bound)]
print(outliers)

print('\nIn some cases, outliers may reflect legitimate extreme values or rare occurrences that are important for the analysis. For instance, when analyzing time use for specific groups (e.g. top earners), these extreme values could provide valuable insights. Therefore, it might be best to retain them in the data-set rather than removing them.')
```
