# Consistency Checks

# 1. Missing Values

```python
# Check for missing values in each column
print(df_whr.isnull().sum())
```

```python
# Create a copy of the dataframe
df_cleaned = df_whr.copy()

# Handle missing values for continuous columns
df_cleaned['Life Ladder'] = df_cleaned['Life Ladder'].fillna(df_cleaned['Life Ladder'].median())
df_cleaned['Log GDP per capita'] = df_cleaned['Log GDP per capita'].fillna(df_cleaned['Log GDP per capita'].median())
df_cleaned['Social support'] = df_cleaned['Social support'].fillna(df_cleaned['Social support'].median())
df_cleaned['Healthy life expectancy at birth'] = df_cleaned['Healthy life expectancy at birth'].fillna(df_cleaned['Healthy life expectancy at birth'].median())
df_cleaned['Freedom to make life choices'] = df_cleaned['Freedom to make life choices'].fillna(df_cleaned['Freedom to make life choices'].median())
df_cleaned['Generosity'] = df_cleaned['Generosity'].fillna(df_cleaned['Generosity'].median())  # Or you could drop rows with too many missing
df_cleaned['Perceptions of corruption'] = df_cleaned['Perceptions of corruption'].fillna(df_cleaned['Perceptions of corruption'].median())
df_cleaned['Positive affect'] = df_cleaned['Positive affect'].fillna(df_cleaned['Positive affect'].median())
df_cleaned['Negative affect'] = df_cleaned['Negative affect'].fillna(df_cleaned['Negative affect'].median())

# After replacing missing values, check if there are any left
print(df_cleaned.isnull().sum())
```

# 2. Inspect Data Types

```python
# Check the data types of all columns
print(df_cleaned.dtypes)
print('-------------------------------------------------')
print('No mixed data type found')
```

# 3. Remove Duplicate Rows if any

```python
# Check for duplicate rows in df_cleaned
duplicate_count = df_cleaned.duplicated().sum()

if duplicate_count == 0:
    print("No duplicates found")
else:
    print(f"Duplicates found: {duplicate_count}")
```

# 4. Check for Consistency in Categorical Data

```python
# Check unique country names for inconsistencies
print(df_cleaned['Country name'].unique())
```

# 5. Creating a new column called 'Happiness Score'

```python
# Create a new 'Happiness Score' in df_cleaned by combining columns (example)
df_cleaned['Happiness Score'] = df_cleaned['Life Ladder'] * df_cleaned['Social support'] / df_cleaned['Perceptions of corruption']
```

# 6. Check for Outliers

```python
# Summary statistics to identify outliers
print(df_cleaned.describe())
```

```python
# Remove rows with a z-score greater than 3 or less than -3
from scipy.stats import zscore

# Calculate z-scores for 'Happiness Score'
df_cleaned['z_score'] = zscore(df_cleaned['Happiness Score'])

# Remove rows where the absolute z-score is greater than 3
df_cleaned_no_outliers = df_cleaned[df_cleaned['z_score'].abs() <= 3]

# Verify by checking the shape before and after removal
print("Original dataframe shape:", df_cleaned.shape)
print("Dataframe shape after removing outliers:", df_cleaned_no_outliers.shape)
```

```python
# Export the dataset as a pickle file

# Define the export path
export_path1 = os.path.join(path, 'Data', 'Prepared Data', 'df_happiness.pkl')

# Save the cleaned dataframe as a pickle file
df_cleaned_no_outliers.to_pickle(export_path1)

print("File saved successfully at:", export_path1)
```
